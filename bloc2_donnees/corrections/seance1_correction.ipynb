### Exercice 2 — Le prix typique

> **Votre mission :**
> - Calculer le prix **moyen** dans `prix_moyen` et le prix **médian** dans `prix_median`.
> - Arrondir les deux à 2 décimales avec `round(..., 2)`.

```python
prix_moyen = round(ventes["prix"].mean(), 2)     ## la moyenne
prix_median = round(ventes["prix"].median(), 2)  ## la valeur du milieu

# Moyenne deux fois plus haute que la mediane : quelques prix tres eleves
# tirent la moyenne vers le haut. C'est la mediane qui decrit le typique.

print("moyenne :", prix_moyen, "| mediane :", prix_median)
```

```python
verifier("2a - prix moyen", prix_moyen == 3.93, "la methode s'appelle mean()")
verifier("2b - prix median", prix_median == 1.95, "la methode s'appelle median()")
```

### Exercice 5 — Le chiffre d'affaires total

> **Votre mission :**

### Exercice 6 — Le pays le plus représenté

> **Votre mission :**
> - Dans `clients`, trouver le pays qui revient le plus souvent.
> - Mettre son **nom** dans `pays_top` et son **effectif** dans `nb_pays_top`.
> - Indice : `value_counts()` trie déjà du plus fréquent au moins fréquent.

```python
# value_counts() renvoie une Series triee par frequence decroissante
compte = clients["pays"].value_counts()

pays_top = compte.index[0]      ## le nom, dans l'index
nb_pays_top = compte.iloc[0]    ## l'effectif, dans les valeurs

# Raccourci utile : compte.idxmax() donne directement l'etiquette du
# maximum, la ou compte.max() donnerait l'effectif.

print(pays_top, ":", nb_pays_top, "clients")
```

```python
verifier("6a - pays le plus represente", pays_top == "Royaume-Uni",
         "value_counts() est deja trie : le premier est le plus frequent")
verifier("6b - effectif", nb_pays_top == 235,
         "iloc[0] donne la premiere valeur, index[0] donne son etiquette")
```

### Exercice 8 — Question de synthèse

> **Votre mission :**
> - Combien de commandes distinctes (`cmd_id`) le fichier contient-il ? → `nb_cmd`
> - Quel est le **panier moyen**, c'est-à-dire le CA total divisé par le nombre de commandes ? → `panier_moyen`, arrondi à 2 décimales.

```python
nb_cmd = ventes["cmd_id"].nunique()   ## commandes distinctes

# Le panier moyen est par COMMANDE, pas par ligne : une commande contient
# souvent plusieurs produits, donc plusieurs lignes.
panier_moyen = round(ca_total / nb_cmd, 2)   ## CA total / nb de commandes

print(nb_cmd, "commandes | panier moyen :", panier_moyen, "euros")
```

### Question 15 — Ce que le catalogue n'est pas

> **Votre mission :**
> - Isoler les références dont le `prod_id` ne contient **aucun chiffre**.
> - Combien y en a-t-il ? Quel chiffre d'affaires représentent-elles, et quelle part du total ?
> - Que faudrait-il faire de ces lignes avant de publier un classement des meilleures ventes ?
> - *Nouveau :* `.str.isalpha()` est vrai quand la chaîne ne contient que des lettres.

```python
faux_produits = produits[produits["prod_id"].str.isalpha()]
print(faux_produits)

# isin() teste l'appartenance a une liste : une condition, 45 123 lignes
lignes = ventes[ventes["prod_id"].isin(faux_produits["prod_id"])]
part = 100 * lignes["ca"].sum() / ventes["ca"].sum()   ## leur poids reel

print(round(lignes["ca"].sum(), 2), "euros, soit", round(part, 2), "% du CA")

# POST = frais de port, M = saisie manuelle. Ce sont des lignes de
# facturation, pas des articles. Les laisser dans un classement produits
# donnerait un "meilleur vendeur" qui n'existe pas. On y revient en 2.4.
```
