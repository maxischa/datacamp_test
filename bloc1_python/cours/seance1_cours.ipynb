## 3. Markdown — écrire proprement autour du code

Les cellules de texte s'écrivent en **Markdown** : quelques signes de
ponctuation qui deviennent de la mise en forme.

| Vous tapez | Vous obtenez |
|---|---|
| `## Mon titre` | un titre |
| `### Sous-titre` | un titre plus petit |
| `**important**` | **important** |
| `*nuance*` | *nuance* |
| `- premier` (une ligne par élément) | une liste à puces |
| `` `code` `` | `code` |
| `> remarque` | une citation en retrait |

> 💡 **Double-cliquez sur cette cellule** pour voir le texte brut qui l'a
> produite, puis `Maj + Entrée` pour la réafficher.

## 4. Vos premières lignes de Python

### Une variable, c'est un nom posé sur une valeur

````python
cout = 5000
revenu = 12500

benefice = revenu - cout
print(benefice)
````

Trois choses viennent de se passer : on a rangé 5 000 dans un nom, puis 12 500
dans un autre, puis calculé avec ces noms plutôt qu'avec les nombres. Si le
coût change, une seule ligne change.

### Chaque valeur a un type

````python
campagne = "Lancement de printemps"   # du texte : un str
budget = 5000                          # un entier : un int
taux = 0.045                           # un decimal : un float

print(type(campagne), type(budget), type(taux))
````

Ça paraît anecdotique. Ça ne l'est pas : `5000` et `"5000"` se ressemblent, et
Python n'en fait pas du tout la même chose.

````python
print(2 * 3)        # avec des nombres : une multiplication
print(2 * "3")      # avec du texte : une repetition
````

`"33"` n'est pas `6`. Le jour où un fichier vous livre des prix en texte — ça
arrivera dès le bloc 2 — vous saurez pourquoi vos totaux sont absurdes.

### Afficher une phrase avec des valeurs dedans

````python
roi = (revenu - cout) / cout

print(f"La campagne a coute {cout} euros et rapporte {revenu} euros.")
print(f"Retour sur investissement : {roi:.1%}")
````

Le `f` devant les guillemets autorise les `{...}` : Python y remplace le nom
par sa valeur. Le `:.1%` demande un pourcentage à une décimale.

C'est la forme que vous utiliserez tout le semestre pour présenter un
résultat.

### L'erreur silencieuse : l'ordre d'exécution

Exécutez la cellule ci-dessous, puis **remontez ré-exécuter la cellule du
bénéfice** (celle avec `revenu - cout`).

````python
cout = 9000      # le budget a ete revu a la hausse
print(cout)
````

Le bénéfice a changé — alors que vous n'avez **rien modifié** dans cette
cellule-là.

Un notebook garde en mémoire ce que vous avez exécuté, **dans l'ordre où vous
l'avez exécuté**, pas dans l'ordre où les cellules sont affichées. C'est le
piège numéro un du notebook, et il ne produit aucun message d'erreur : juste
des résultats faux.

> ⚠️ **Le réflexe :** en cas de doute, *Exécution → Redémarrer et tout
> exécuter*. Tout est recalculé de haut en bas, dans l'ordre visible.
