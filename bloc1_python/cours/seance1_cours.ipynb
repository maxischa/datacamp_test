## 4. Vos premières lignes de Python

### Une variable, c'est un nom posé sur une valeur

````python
cout = 5000
revenu = 12500

benefice = revenu - cout
print(benefice)
````

Trois choses viennent de se passer : on a rangé 5 000 dans un nom, puis 12 500

### Chaque valeur a un type

````python
campagne = "Lancement de printemps"   # du texte : un str
budget = 5000                          # un entier : un int
taux = 0.045                           # un decimal : un float

print(type(campagne), type(budget), type(taux))
````

Ça paraît anecdotique. Ça ne l'est pas : `5000` et `"5000"` se ressemblent, et
Python n'en fait pas du tout la même chose.

````python
print(2 * 3)        # avec des nombres : une multiplication
print(2 * "3")      # avec du texte : une repetition
````

`"33"` n'est pas `6`. Le jour où un fichier vous livre des prix en texte — ça

### Afficher une phrase avec des valeurs dedans

````python
roi = (revenu - cout) / cout

print(f"La campagne a coute {cout} euros et rapporte {revenu} euros.")
print(f"Retour sur investissement : {roi:.1%}")
````

Le `f` devant les guillemets autorise les `{...}` : Python y remplace le nom

## 6. Décider : `if` / `else`

Jusqu'ici, votre code faisait toujours la même chose. Une décision de gestion,
elle, dépend du résultat : *« reconduit-on la campagne ? »* — ça dépend du ROI.

````python
roi = 0.35
seuil = 0.20   # en dessous, le comite ne reconduit pas

if roi > seuil:
    print("Campagne reconduite")
else:
    print("Campagne arretee")
````

Deux choses à retenir de cette forme :

- la ligne de test finit par **deux-points** ;
- ce qui suit est **décalé vers la droite** de quatre espaces. C'est ce décalage
  qui dit à Python « ceci ne s'exécute que si le test est vrai ».

> 📱 Sur tablette, ne comptez pas les espaces : Colab décale tout seul quand
> vous passez à la ligne après les deux-points.

### Plus de deux cas : `elif`

````python
if roi > 0.50:
    verdict = "on augmente le budget"
elif roi > 0.20:
    verdict = "on reconduit a l'identique"
else:
    verdict = "on arrete"

print(verdict)
````

Python lit les cas **de haut en bas** et s'arrête au premier qui est vrai. Un
ROI de 0,60 déclenche la première ligne, et la deuxième n'est même pas
regardée.

### Comparer : les six signes

| | |
|---|---|
| `>` `<` | strictement supérieur, strictement inférieur |
| `>=` `<=` | supérieur ou égal, inférieur ou égal |
| `==` | **est égal à** |
| `!=` | est différent de |

Le seul qui pose problème est `==`.

````python
if roi = 0.35:
    print("jamais affiche")
````

Dernière ligne du message :

```
SyntaxError: invalid syntax. Maybe you meant '==' or ':=' instead of '='?
```

Python vous donne ici la correction lui-même — ça n'arrive pas souvent, mais
ça arrive : **lisez la dernière ligne avant de chercher ailleurs.**

`=` **pose** une valeur (`roi = 0.35`). `==` **demande** si deux valeurs sont
égales (`roi == 0.35`). Une seule touche de différence, et c'est la faute la
plus fréquente des trois premières séances.

## 7. Demander une valeur : `input()`

Vos chiffres sont écrits en dur dans le code. Pour tester un autre scénario, il
faut modifier une ligne — et si c'est un collègue qui veut essayer, il faut lui
expliquer laquelle.

`input()` affiche une question, attend une réponse, et la range dans une
variable. Votre notebook devient un petit **simulateur**.

````python
cout_texte = input("Cout de la campagne en euros : ")

cout = float(cout_texte)   # input rend TOUJOURS du texte : on convertit
revenu = 12000

roi = (revenu - cout) / cout
print(f"Pour {cout:.0f} euros investis, ROI = {roi:.1%}")
````

> ⚠️ **Le point unique de cette section : `input()` renvoie toujours du
> texte**, même quand vous tapez `5000`. Sans le `float(...)`, la soustraction
> `revenu - cout` échouerait sur un `TypeError`.

`float(...)` pour un décimal, `int(...)` pour un entier — c'est la conversion
que vous avez déjà vue au paragraphe 4, cette fois rendue obligatoire.

> 📱 Le champ de saisie apparaît **sous la cellule**, et le clavier vous masque
> la moitié de l'écran. Tapez, validez par `Entrée`, le clavier disparaît. Tant
> que vous n'avez pas validé, la cellule reste en attente et rien d'autre ne
> s'exécute.

## 8. Répéter : la boucle `for`

Une seule valeur de coût ne suffit pas : le comité veut voir **plusieurs
scénarios**. Vous pourriez recopier le calcul quatre fois. Ne le faites pas.

````python
couts = [4000, 6000, 8000, 10000]
revenu = 12000

for cout in couts:
    roi = (revenu - cout) / cout
    print(f"cout {cout} euros -> ROI {roi:.1%}")
````

La variable `cout` prend **chaque valeur de la liste à tour de rôle**, et le
bloc décalé s'exécute une fois par valeur. Cinq lignes de code, quatre
scénarios ; il y en aurait cinquante que le code serait le même.

### Garder les résultats plutôt que les afficher

Afficher, c'est bien pour regarder. Pour **tracer un graphique**, il faut
ranger les résultats au fur et à mesure.

````python
rois = []   # une liste vide, qu'on remplit au fil de la boucle

for cout in couts:
    rois.append((revenu - cout) / cout)   # append ajoute a la fin

print(rois)
````

`liste.append(valeur)` ajoute un élément à la fin. À la sortie de la boucle,
`rois` contient les quatre résultats, prêts à être tracés.

> 💡 `range(2000, 20001, 2000)` fabrique une suite de nombres — de 2 000 à
> 20 000 par pas de 2 000 — sans que vous ayez à les taper. `list(range(...))`
> la transforme en liste. Vous en aurez besoin dans la feuille d'exercices.
