> **Votre mission :**

## 7. Décider : `if` / `else`

Jusqu'ici, votre code faisait toujours la même chose. Une décision de gestion,
elle, dépend du résultat : *« reconduit-on la campagne ? »* — ça dépend du ROI.

````python
roi = 0.35
seuil = 0.20   # en dessous, le comite ne reconduit pas

if roi > seuil:
    print("Campagne reconduite")
else:
    print("Campagne arretee")
````

Deux choses à retenir de cette forme :

- la ligne de test finit par **deux-points** ;
- ce qui suit est **décalé vers la droite** de quatre espaces. C'est ce décalage
  qui dit à Python « ceci ne s'exécute que si le test est vrai ».

> 📱 Sur tablette, ne comptez pas les espaces : Colab décale tout seul quand
> vous passez à la ligne après les deux-points.

### Plus de deux cas : `elif`

````python
if roi > 0.50:
    verdict = "on augmente le budget"
elif roi > 0.20:
    verdict = "on reconduit a l'identique"
else:
    verdict = "on arrete"

print(verdict)
````

Python lit les cas **de haut en bas** et s'arrête au premier qui est vrai. Un
ROI de 0,60 déclenche la première ligne, et la deuxième n'est même pas
regardée.

### Comparer : les six signes

| | |
|---|---|
| `>` `<` | strictement supérieur, strictement inférieur |
| `>=` `<=` | supérieur ou égal, inférieur ou égal |
| `==` | **est égal à** |
| `!=` | est différent de |

Le seul qui pose problème est `==`.

````python
if roi = 0.35:
    print("jamais affiche")
````

Dernière ligne du message :

```
SyntaxError: invalid syntax. Maybe you meant '==' or ':=' instead of '='?
```

Python vous donne ici la correction lui-même — ça n'arrive pas souvent, mais
ça arrive : **lisez la dernière ligne avant de chercher ailleurs.**

`=` **pose** une valeur (`roi = 0.35`). `==` **demande** si deux valeurs sont
égales (`roi == 0.35`). Une seule touche de différence, et c'est la faute la
plus fréquente des trois premières séances.

## 8. Répéter : la boucle `for`

Une seule valeur de coût ne suffit pas : le comité veut voir **plusieurs
scénarios**. Vous pourriez recopier le calcul quatre fois. Ne le faites pas.

````python
couts = [4000, 6000, 8000, 10000]
revenu = 12000

for cout in couts:
    roi = (revenu - cout) / cout
    print(f"cout {cout} euros -> ROI {round(roi * 100)} %")
````

La variable `cout` prend **chaque valeur de la liste à tour de rôle**, et le
bloc décalé s'exécute une fois par valeur. Cinq lignes de code, quatre
scénarios ; il y en aurait cinquante que le code serait le même.

### Garder les résultats plutôt que les afficher

Afficher, c'est bien pour regarder. Pour **tracer un graphique**, il faut
ranger les résultats au fur et à mesure.

````python
rois = []   # une liste vide, qu'on remplit au fil de la boucle

for cout in couts:
    rois.append((revenu - cout) / cout)   # append ajoute a la fin

print(rois)
````

`liste.append(valeur)` ajoute un élément à la fin. À la sortie de la boucle,
`rois` contient les quatre résultats, prêts à être tracés.

> 💡 `range(2000, 20001, 2000)` fabrique une suite de nombres — de 2 000 à
> 20 000 par pas de 2 000 — sans que vous ayez à les taper. `list(range(...))`
> la transforme en liste. Vous en aurez besoin dans la feuille facultative.

> - Ranger les trois budgets 1500, 2300 et 900 dans une liste `budgets`.
> - Compter les campagnes → `nb_budgets`, et faire le total des budgets → `total_budgets`. **Avec la boucle `for` que vous venez de voir**, pas autrement.
> - Le principe : deux variables parties de **zéro** avant la boucle — un **compteur**, qui gagne 1 à chaque tour, et un **accumulateur**, qui reçoit le budget du tour.

````python
# Une liste : des valeurs entre crochets, separees par des virgules
budgets = [1500, 2300, 900]

# Les deux partent de zero AVANT la boucle : a l'interieur, elles seraient
# remises a zero a chaque tour et n'accumuleraient rien
nb_budgets = 0       # le compteur
total_budgets = 0    # l'accumulateur

for budget in budgets:
    nb_budgets = nb_budgets + 1              ## une campagne de plus
    total_budgets = total_budgets + budget   ## on empile les euros

print(nb_budgets, "campagnes pour", total_budgets, "euros")
````

````python
verifier("nombre de budgets", nb_budgets == 3,
         "le compteur part de 0 et gagne 1 a chaque tour de boucle")
verifier("total des budgets", total_budgets == 4700,
         "l'accumulateur part de 0 et recoit le budget du tour : 1500 + 2300 + 900")
````

> **Votre mission :**

> - Mettre le **numéro** de la meilleure (1, 2 ou 3) dans `meilleure`.
> - *Nouveau :* `couts[i]` donne le i-ème élément d'une liste, et `range(3)` fabrique 0, 1, 2.

````python
couts = [1500, 2300, 900]
revenus = [3000, 2100, 2700]

for i in range(3):
    benefice = revenus[i] - couts[i]
    roi = round(100 * benefice / couts[i])   ## en pourcentage
    print(f"Campagne {i + 1} : {benefice} euros de benefice, ROI {roi} %")

meilleure = 3   ## 900 euros investis, 1 800 de benefice : 200 % de ROI

# La deuxieme est la seule a perdre de l'argent — et c'est pourtant celle
# qui a le plus gros budget. Le montant investi ne dit rien du rendement.
````

### Question 7 — À partir de quel coût la campagne perd-elle de l'argent ?

> **Votre mission :**
> - Le revenu attendu est de **12 000 €**, quoi qu'il arrive. Seul le coût est encore négociable.
> - Calculer le ROI pour tous les coûts de **2 000 à 20 000 € par pas de 2 000**, avec une boucle `for`.
> - Tracer la courbe du ROI en fonction du coût, et y ajouter la ligne du **ROI nul** : `plt.axhline(0, color="red")`.
> - Répondez en commentaire : **à partir de quel coût le ROI devient-il négatif**, et pourquoi ce point-là ?
> - *Nouveau :* `list(range(2000, 20001, 2000))` fabrique la liste des coûts, `plt.plot(x, y)` trace la courbe.

````python
revenu = 12000
couts = list(range(2000, 20001, 2000))
rois = []

for cout in couts:
    rois.append((revenu - cout) / cout)

plt.figure(figsize=(7, 4))
plt.plot(couts, rois, marker="o")
plt.axhline(0, color="red")   # le seuil : au-dessous, on perd de l'argent
plt.title("ROI selon le cout, a revenu constant (12 000 euros)")
plt.xlabel("cout de la campagne (euros)")
plt.ylabel("ROI")
plt.show()

# Le ROI s'annule EXACTEMENT a 12 000 euros, et devient negatif des 14 000.
# Ce n'est pas un hasard : le ROI vaut (revenu - cout) / cout, donc il change
# de signe quand le cout depasse le revenu. Le graphique montre en plus ce
# qu'un tableau cache — la courbe s'effondre a gauche : les 6 000 premiers
# euros font presque tout le ROI, les suivants n'achetent plus grand-chose.
````
