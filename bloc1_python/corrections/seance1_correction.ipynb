### Question 12 — Le rapport complet d'une campagne

> **Votre mission :**
> - Une campagne email : 20 000 envois, 500 € de coût, 22 % d'ouvertures, 3 % de clics parmi les ouvertures, 18 € de panier moyen par clic.
> - Calculer le nombre d'ouvertures, le nombre de clics, le revenu, le bénéfice et le ROI.
> - Afficher un rapport de cinq lignes avec des `f-strings`.

```python
envois = 20000
cout = 500
taux_ouverture = 0.22
taux_clic = 0.03
panier = 18

ouvertures = envois * taux_ouverture
clics = ouvertures * taux_clic
revenu = clics * panier
benefice = revenu - cout

print(f"Envois      : {envois}")
print(f"Ouvertures  : {ouvertures:.0f}")
print(f"Clics       : {clics:.0f}")
print(f"Revenu      : {revenu:.2f} euros")
print(f"Benefice    : {benefice:.2f} euros ({benefice / cout:.0%} de ROI)")
```

### Question 13 — Comparer trois campagnes

> **Votre mission :**
> - Trois campagnes ont coûté 1 500, 2 300 et 900 € et rapporté 3 000, 2 100 et 2 700 €.
> - Ranger coûts et revenus dans deux listes, puis afficher pour chacune son bénéfice.
> - Laquelle a le meilleur ROI ? Répondez en commentaire.

```python
couts = [1500, 2300, 900]
revenus = [3000, 2100, 2700]

for i in range(3):
    benefice = revenus[i] - couts[i]
    print(f"Campagne {i + 1} : {benefice} euros de benefice, "
          f"ROI {benefice / couts[i]:.0%}")

# La troisieme : 900 euros investis, 1 800 de benefice, soit 200 % de ROI.
# La deuxieme est la seule a perdre de l'argent — et c'est pourtant celle
# qui a le plus gros budget.
```

### Question 15 — Formater proprement un montant

> **Votre mission :**
> - Afficher le montant `1234.5678` arrondi à 2 décimales, puis sans décimale ; et le taux `0.0456` en pourcentage à 1 décimale.
> - *Nouveau :* dans un f-string, `{valeur:.2f}` arrondit, `{valeur:.1%}` met en pourcentage, `{valeur:.0f}` supprime les décimales.

```python
montant = 1234.5678
taux = 0.0456

print(f"deux decimales : {montant:.2f} euros")
print(f"sans decimale  : {montant:.0f} euros")
print(f"en pourcentage : {taux:.1%}")

# Un rapport de gestion n'affiche jamais 1234.5678. Et le format
# pourcentage s'applique a un TAUX (0,0456), pas a un montant : sur
# 1234.5678 il afficherait 123456,8 %.
```
